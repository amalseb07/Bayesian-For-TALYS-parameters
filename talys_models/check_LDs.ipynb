```python
import pandas as pd 
import numpy as np

import matplotlib.pyplot as plt
plt.rcParams['figure.figsize']=(14.4,7.4)

```

```python
#ld model 5 positive parity

df1=pd.read_table("ld_random_p.txt",delimiter=r"\s+")

df1['ld_eff']=df1['JP=0.5']+df1['JP=1.5']+df1['JP=2.5']

df1=df1[df1['Ex']<8]    


df1
```

```python
df2=pd.read_table("ld_random_n.txt",delimiter=r"\s+")

df2['ld_eff']=df2['JP=0.5']+df2['JP=1.5']+df2['JP=2.5']

df2=df2[df2['Ex']<8]    

df2

```

```python
df = df1.set_index('Ex').add(df2.set_index('Ex'), fill_value=0).reset_index()


df=df[df['Ex']>0.25]  

df


#df.to_csv("C:\\Users\\amals\\bayesian-ld\\ld_300(6,9)_talys.csv", index=False)

df['ld_eff']
```

```python
# construct

df_p =pd.read_pickle("master_base_ld_p.pkl")
df_n = pd.read_pickle("master_base_ld_n.pkl")


df_base_p =pd.read_pickle("base_ld_p.pkl")
df_base_n = pd.read_pickle("base_ld_n.pkl")

ctable = -0.08
ptable = 0.4 

new_energy = df_base_p['Ex'].values - ptable

# for postive_ld
for col in df_base_p.columns[1:]:
    shifted_values = np.interp(new_energy, df_p['Ex'].values, df_p[col].values)
    
    new_values = np.exp(ctable* np.sqrt(np.abs(new_energy)))* shifted_values
    
    df_base_p[col] = new_values

new_energy = df_base_n['Ex'].values - ptable

# for postive_ld
for col in df_base_n.columns[1:]:
    shifted_values = np.interp(new_energy, df_n['Ex'].values, df_n[col].values)
    
    new_values = np.exp(ctable* np.sqrt(np.abs(new_energy)))* shifted_values
    
    df_base_n[col] = new_values


df_final= df_base_p.set_index('Ex').add(df_base_n.set_index('Ex'), fill_value=0).reset_index()

df_final['ld_eff']=df_final['JP=0.5']+df_final['JP=1.5']+df_final['JP=2.5']

#df_final=df_final[df_final['Ex']>0.25]  

#df_final=df_final[df_final['Ex']<8]  
df_p
```

```python
#plot both together 

fig = plt.figure()

#plt.plot(df1['energy'], df1['ld'],color='b',label='ld1 model')
plt.plot(df['Ex'], df['ld_eff'],color='green',label='ld-model5-talys')

plt.plot(df_final['Ex'], df_final['ld_eff'],color='blue',label='construct')

#plt.errorbar(dfd_['energy'], dfd_['ld'],yerr= dfd_['errors'],fmt="o", color="r",ecolor='r',linestyle='None',label='97 Zr(oslo)')

plt.yscale("log")

plt.xlabel("Excitation Energy(MeV)",fontsize = 20)
plt.ylabel("Level density (MeV$^{-1}$)",fontsize = 20) 
plt.title("Level density",fontsize = 20)    

plt.xlim(0, 6)
plt.xticks(fontsize = 20)
plt.yticks(fontsize = 20)

#plt.ylim(10**(-9),10**(-4))

plt.legend(loc ='lower right',fontsize="x-large")
plt.show()
```
