```python
import pandas as pd 
import numpy as np

import matplotlib.pyplot as plt
plt.rcParams['figure.figsize']=(14.4,7.4)

```

```python
df8=pd.read_table("C:\\Users\\amals\\OneDrive\\Desktop\\ANL_e1807_with_97Y_emphasis\\talys\\what is talys actually doing to gsf\\new_best_talys for 97Zr\\gsf.txt",delimiter="\s+",skiprows=2,header=None)

#df8=pd.read_table("C:\\Users\\amals\\Downloads\\randomgsf.txt",delimiter="\s+",skiprows=2,header=None)
df8.columns=['energy','f(M1)','f(E1)','T(M1)','T(E1)']
df8['gsf'] = df8['f(M1)'] + df8['f(E1)']

df8.head(30)


#df8.to_csv("C:\\Users\\amals\\bayesian\\gsf_300(6,9)_talys.csv", index=False)


df8.head(50)
```

```python
# read file from strength.nrm that is saved in following directory

dfd_=pd.read_table("C:\\Users\\amals\\OneDrive\\Desktop\\anchor game complete\\no_isomer_ld\\6and9\\gsf.txt")

dfl, dfu =  np.array_split(dfd_, 2)
dfl.insert(1, "errors", dfu["gsf"].values)

#energy array created
a0=-0.7590                    #from strength.cpp file constants need to be extracted
a1= 0.300

l=len(dfl)

energy=np.zeros(l)

for i in range(l):
    energy[i]= a0+a1*i 

#adding energy to data frame

dfl['energy'] = energy.tolist()

# changing energy to be first column
first_column = dfl.pop('energy') 
dfl.insert(0, 'energy', first_column)


#extracting columns with positive gsf    
    
dfd_=dfl[dfl['gsf']>0]    
    
    
dfd_=dfd_[dfd_['gsf']>dfd_['errors']]




#dfd_.to_csv("C:\\Users\\amals\\bayesian\\gsf_300(6,9).csv", index=False)

dfd_
```

```python
df_construct = pd.read_pickle("master_base_gsf.pkl")

# f(E1)
ftable=1
C= 10**(-10)
U =  5.569156
eta =3

#f(M1)
upbende =1.1
upbendc = 1.5*10**(-7)

# f(e1) construction
df_construct['f(E1)']= ftable * df_construct['f(E1)']+  (C* (U-df_construct['energy']))/(1+np.exp(df_construct['energy']-eta))

#f(m1) construction
df_construct['f(M1)']=  df_construct['f(M1)']+  upbendc * np.exp(-1*upbende*df_construct['energy'])

df_construct['gsf'] = df_construct['f(E1)']+ df_construct['f(M1)']


df_construct.head(30)
```

```python
#plot both together 

fig = plt.figure()


plt.plot(df8['energy'], df8['gsf'],color='forestgreen',label='strength-model8-talys')

plt.plot(df_construct['energy'], df_construct['gsf'],color='blue',label='construct')

#plt.plot(dfd_['energy'], dfd_['gsf'],color='r',marker='*',linestyle='None')
plt.errorbar(dfd_['energy'], dfd_['gsf'],yerr= dfd_['errors'],fmt="o", color="r",ecolor='r',label='97 Zr(oslo)')

plt.yscale("log")

plt.xlabel("Gamma energy(MeV)",fontsize = 20)
plt.ylabel("GSF(MeV$^{-3}$)",fontsize = 20) 
plt.title("γ-ray strength function",fontsize = 20)    

plt.xticks(fontsize = 20)
plt.yticks(fontsize = 20)


plt.xlim(0, 20)

plt.ylim(10**(-9),10**(-4))

plt.legend(loc ='lower right',fontsize="x-large")
plt.show()
```
